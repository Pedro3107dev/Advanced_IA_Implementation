Librerias usadas:

```python
import random
import matplotlib.pyplot as plt
```

Función para simular slots machines

```python
def rewardsList(k, possibleRewards, probabilities):
    rk = []
    for i in range(k):
        rktemp = random.choices(possibleRewards, probabilities)[0]
        rk.append(rktemp)

    return rk
```

Función para calcular el valor esperado "expected value"

```python
def expected_value(k, rk):
    Qk = [0, 0]
    for i in range(k):
        qk = Qk[i - 1] + 1 / k * (rk[i] - Qk[i - 1])
        Qk.append(qk)

    return Qk
```

Función para proyectar variación de resultados del expected value según la iteración en k

```python
def  plotExpectedValue(Qk1, Qk2, k):
    plt.plot(range(k + 2), Qk1, label = 'Slot Machine 1')
    plt.plot(range(k + 2), Qk2, label = 'Slot Machine 2')
    plt.xlabel('Number of Plays (k)')
    plt.ylabel('Expected Value (Qk)')
    plt.title(f'Expected Value Comparison for k = {k}')
    plt.legend()
    plt.show()
```

Sección 'main' del código en la que se inicializa los valores de k en los que se va a iterar. Se llaman las tres funciones anteriores para simular los premios dados por las "slot machines", se calcula el "expected value", se imprimen los resultados en la terminal y se presenta en una gráfica seguún el valor de 'k' y lsus respectivos valores del "expected value".

```python
# Valores
# k = 100, 1000, 10000
K = [100, 1000, 10000]

# Probabilidades y recompensas de las slot machines:
rewardsSlotMachine1 = [0, 10]
prob1 = [0.6, 0.4]
rewardsSlotMachine2 = [0, 100]
prob2 = [0.97, 0.03]

for i in range(len(K)):
    # Número de jugadas | Iterando entre (10, 100 y 1000 iteraciones)
    k = K[i]

    # Simulación de juego de las slot machines k veces
    rk1 = rewardsList(k, rewardsSlotMachine1, prob1)
    rk2 = rewardsList(k, rewardsSlotMachine2, prob2)
    # print(f"Resultados de rk1: {rk1}")
    # print(f"Resultados de rk2: {rk2}")

    # Cálculo del valor esperado Qk para cada slot machine:
    Qk1 = expected_value(k, rk1)
    Qk2 = expected_value(k, rk2)

    print(f"\n---Learning rate: 1/{k}---")
    print(f"\nValor esperado de la slot machine 1:")
    print(Qk1)
    print(f"\nValor esperado de la slot machine 2:")
    print(Qk2)
    print("\n")

    # Gráfica de los valores esperados
    plotExpectedValue(Qk1, Qk2, k)
```
